Today, we'll look at metrics. **A metric allows us to state precisely how well our model is performing.**

> We've already seen a metric - the Sum of Squared Errors (SSE)!

You can consider a metric to be a distance function -- it measures how far away our predictions are from the actual targets.

### What is a metric?

Metric is really short for **distance metric**. It is used to give us a numerical measure of how distant two things are.

Metrics actually have a strict mathematical definition. Given two _vectors_ $x, y$, a metric must satisfy four rules:

###### Rule 1: Distances must be positive
$$d(x, y) \ge 0$$

###### Rule 2: If a distance between two things is zero, then they're the same thing
$$d(x, y) = 0 \iff x = y$$

###### Rule 3: Distances are symmetric
$$d(x, y) = d(y, x)$$

###### Rule 4: The Triangle Inequality
The distance from $x$ to $y$ must be shorter than going through a new point ($z$) in between.
$$d(x, y) \le d(x, z) + d(z, y)$$

### Why do we care?
The goal of our model is to get our predictions $\hat{\mathbf{y}}$ (a vector!) as close as possible to the actual truth $\mathbf{y}$ (also a vector!). ie, **we want to minimize some distance metric between the truth and our predictions!**

### The "Usual" Distance Metric
The shortest coordinate-plane distance between two points $\mathbf{x} = (x_1, x_2)$ and $\mathbf{y} = (y_1, y_2)$ is calculated as:

$$d(\mathbf{x}, \mathbf{y}) = \sqrt{ (y_1 - x_1)^2 + (y_2 - x_2)^2 }$$

This has a fancy name that we might use later: **Euclidean distance** or **$\mathcal{l}_2$-distance.**

This can generalize to $n$-dimensional vectors. That is, for $\mathbf{x} = (x_1, ..., x_n)$ and $\mathbf{y} = (y_1, ..., y_n)$, then

$$d(\mathbf{x}, \mathbf{y}) = \sqrt{\sum (y_i - x_i)^2}$$

### Some Fancy Notation
The distance from a vector $\mathbf{x}$ to the origin, aka the **norm** or **magnitude** of that vector is denoted $\|\mathbf{x}\|$. The distance between two vectors $\mathbf{x}$, $\mathbf{y}$ can also be computed:

$$\|\mathbf{y} - \mathbf{x}\|$$

---

### What is a Loss Function?
A **loss function** is a **distance metric** that describes how far our predictions ($\mathbf{\hat{y}}$) are from the ground truth ($\mathbf{y}$). We seek to minimize these loss metrics. Most ML algorithms are actually fit by finding the best set of parameters that _minimize a specific loss metric_.

### Terminology Disambiguation
A _loss function_ (aka, a _distance metric_ ) computes a _loss metric_ or _error metric_. The process of computing a loss function is sometimes called _scoring_. If the loss function was also used to fit the model, then it's also the model's _objective function_. 

But don't worry about this... everyone just kinda uses all of these interchangably in practice.

# Metric #2: Sum Squared Error (SSE)
The SSE is the the sum of the squares of the errors (ie, residuals) of your model. It's computed as follows:

$$SSE = \sum (y_i - \hat{y}_i)^2 = \|\mathbf{y} - \hat{\mathbf{y}}\|^2$$

**(THREAD):** We'll see that the SSE is not human interpretable. Why would we still care about it?

#### SSE Advantages:
- It's literally what OLS uses to fit the best model anyway!
- It's foundational for computing other error metrics below
- You can still compare SSEs between models (lower is better)
- Punishes "big" residuals more harshly
- It's DIFFERENTIABLE! <- Big deal later!

#### SSE Disadvantages:
- Not human interpretable
- Punishes "big" residuals more harshly
- Units are square

```python
# Compute the SSE using numpy
sse = np.sum((y - y_hat)**2)
sse
```

#### What do we mean OLS "uses" SSE to fit?
Let's say we have the simple linear regression model:

$$\hat{y}_i = \hat{\beta}_0 + \hat{\beta}_1x_i + e_i$$

And so the SSE is:

$$SSE = \sum (y_i - \hat{y})^2 = \sum (y_i - \hat{\beta}_0 - \hat{\beta}_1x_i)^2$$

The act of fitting OLS is the act of finding the _best_ values for $\hat\beta_0$ and $\hat\beta_1$ that make SSE the smallest!

# Metric #3: Mean Squared Error (MSE)
The MSE is simply the _mean_ of the squared errors (ie, residuals). It's computed:

$$MSE = \frac{1}{n}\sum (y_i - \hat{y}_i)^2 = \frac{1}{n}SSE$$

#### MSE Advantages:
- Same as for SSE!
- _Slightly_ human interpretable

#### MSE Disadvantages:
- Still technically not _really_ human interpretable

```python
# Compute the MSE using numpy
mse = np.mean((y - y_hat)**2)
mse
```

```python
# Compute the MSE using sklearn!
metrics.mean_squared_error(y, y_hat)
```

### Metric #3.5: The Root Mean Squared Error (RMSE)
If the MSE is sort of like a variance, we might like to also look at its square root, which would then be sort of like a standard deviation:

$$RMSE = \sqrt{MSE}$$

```python
# Compute the RMSE! Does it make sense?
np.sqrt(metrics.mean_squared_error(y, y_hat))
```

# Metric #4: The Coefficient of Determination ($R^2$)
One of the most well-liked regression metrics is the $R^2$, aka the coefficient of determination. Its interpretation is easy to understand, but _why_ it has that interpretation can be difficult. First, the formula:

$$R^2 = 1 - \frac{SSE}{SSTO}$$

where

$$SSTO = \sum (y_i - \bar{y})^2$$

#### $R^2$ Advantages:
- Interpretable
- On a common scale
- Easy to explain to anyone

#### $R^2$ Disadvantages:
- Interpretation only valid for OLS

```python
# Let's compute it "by hand"
sse = np.sum((y - y_hat)**2)
ssto = np.sum((y - np.mean(y))**2)
1 - sse / ssto
```

```python
# Can also compute directly via sklearn!
lm.score(X, y)
```

```python
# ...and via metrics
metrics.r2_score(y, y_hat)
```

### $R^2$ interpretation:
> $R^2$% of the variability in $y$ can be explained by $X$.

**NB:** This interpretation is only valid for linear regressions! We compute $R^2$ for other types of models, but they don't enjoy the same interpretation.

### Let's interpret:
89.7% of the variability in sales can be explained by advertising in TV, newspaper, and radio.

## More Discussion: What is $SSTO$, exactly?
$$SSTO = \sum (y_i - \bar{y})^2$$

What is the absolute _worst_ model we can fit? It's the one with no x-variables! If we have no x-variables, the best prediction for $y$ would simply be the mean of the $y$s we have, ie, $\bar{y}$.

The model that predicts $\bar{y}$ for every observation is known as the **null** model, and is often used as a benchmark against which we measure our model.

If you look closely, $SSTO$ is just the $SSE$ of the null model!

```python
# Let's compute the SSE, MSE, RMSE of the null model.
m = y.mean()
sse = np.sum((y - m)**2)
mse = np.mean((y - m)**2)
rmse = np.sqrt(mse)

print(f"SSE = {sse}")
print(f"MSE = {mse}")
print(f"RMSE = {rmse}")
```

```python
# Compare to our OLS model:
sse = np.sum((y - y_hat)**2)
mse = metrics.mean_squared_error(y, y_hat)
rmse = np.sqrt(mse)

print(f"SSE = {sse}")
print(f"MSE = {mse}")
print(f"RMSE = {rmse}")
```

# Metric #5: Mean Absolute Deviation
This metric makes the most "sense" to a human, but has flaws:

$$MAE = \frac{1}{n} \sum | y_i - \hat{y}_i |$$

This represents, on average, how far your predictions are from the truth.

#### MAE Advantages
- Interpretable
- On an understandable scale
- Robust to outliers

#### MAE Disadvantages
- Not differentiable
- Might not show off the "best" model
- Punishes all residuals equally

```python
# Compute!
metrics.mean_absolute_error(y, y_hat)
```

# Other Metrics
The art of selecting a metric is determined by what you want to show, and to whom. It's not rare to invent a new one depending on your subject matter. 

There are also a ton more than we won't discuss. These are only valid for OLS, and their goal is to penalize the metric if too many x-variables are used. We have better tools for this now, so these are considered a bit old-fashioned. They're not even included with `sklearn`!

#### Old-fashioned metrics:
- Mallows's $C_p$
- Akaike Information Criterion (AIC)
- Akaike Information Criterion correction (AICc)
- Bayesian Information Criterion (BIC)
